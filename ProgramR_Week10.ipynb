## 1. Strings – Substitution and Replacement

The lecture explains that `sub()` replaces the **first** matching occurrence, while `gsub()` performs a **global replacement** and replaces all matching occurrences.

````r
# Lecture example: sub()
y <- "Number of participants: 25"

y
sub("25", "30", y)
````

````r
# Another replacement
y <- "Number of participants: 25"

sub("participants", "students", y)
````

````r
# Demonstrating that sub() changes only the first occurrence
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."

y
sub("Mr. Singh", "Professor Jha", y)
````

````r
# gsub() replaces ALL occurrences
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."

gsub("Mr. Singh", "Professor Jha", y)
````

### `sub()` vs `gsub()` — important for exam

| Function | Replacement |
|---|---|
| `sub(old, new, string)` | First occurrence only |
| `gsub(old, new, string)` | All occurrences |

````r
# Side-by-side comparison
sentence <- "R is easy. R is powerful. R is useful."

cat("Original:\n", sentence, "\n\n")
cat("sub():\n", sub("R", "R Programming", sentence), "\n\n")
cat("gsub():\n", gsub("R", "R Programming", sentence), "\n")
````

# 2. `grep()` – Searching for Matches

`grep()` searches for a pattern inside each element of a character vector.

By default:
- `value = FALSE` → returns **indices**
- `value = TRUE` → returns the **matching strings**

````r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)

str
````

````r
# Lecture example: value = TRUE
grep("ex", str, value = TRUE)
````

````r
# Lecture example: value = FALSE
grep("ex", str, value = FALSE)
````

### `grep()` with `ignore.case`

````r
str <- c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)

# Case-sensitive search
grep("R", str, ignore.case = FALSE, value = TRUE)
````

````r
# Case-insensitive search
grep("R", str, ignore.case = TRUE, value = TRUE)
````

````r
# Case-sensitive search returning indices
grep("R", str, ignore.case = FALSE, value = FALSE)
````

````r
# Case-insensitive search returning indices
grep("R", str, ignore.case = TRUE, value = FALSE)
````

### Important exam point

`ignore.case = FALSE` means uppercase/lowercase matter.

`ignore.case = TRUE` means uppercase/lowercase are ignored while matching.

## 3. `grep()` with Combined Strings

````r
x <- "R course 24.07.2021"
y <- "Number of participants: 25"

# Combine the two strings
combined <- c(x, y)
combined
````

````r
# Search for "our"
grep("our", c(x, y))
````

`"our"` occurs in `"course"`, which is the **first element**, so the result is `1`.

````r
# Search for "Num"
grep("Num", c(x, y))
````

`"Num"` occurs in `"Number"` in the **second element**, so the result is `2`.

## 4. `grepl()` – TRUE/FALSE Matching

Unlike `grep()`, `grepl()` returns a logical vector:

- `TRUE` → pattern is present
- `FALSE` → pattern is absent

````r
str <- c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grepl("R", str)
````

````r
grepl("ex", str)
````

### `grepl()` with case sensitivity

````r
str <- c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)

# Case-sensitive
grepl("R", str, ignore.case = FALSE)
````

````r
# Case-insensitive
grepl("R", str, ignore.case = TRUE)
````

## 5. Quick String Practice

Run these to reinforce the four major string functions.

````r
texts <- c(
  "R programming is useful",
  "Python programming is also useful",
  "R is used for statistics",
  "Data analysis is important"
)

# Find elements containing "R"
grep("R", texts, value = TRUE)

# Find their indices
grep("R", texts)

# TRUE/FALSE for each element
grepl("R", texts)

# Replace only first matching "programming"
sub("programming", "coding", texts)

# Replace every matching "programming"
gsub("programming", "coding", texts)
````

## 8. `summary()` on a Data Frame

````r
summary(painters)
````

For numerical variables, `summary()` displays:

- Minimum
- 1st Quartile
- Median
- Mean
- 3rd Quartile
- Maximum

For the categorical `School` variable, it displays frequencies.

The lecture notes that only the most frequent categories are displayed by default, with less frequent categories combined as `Other`.

````r
# Summary of a single categorical variable
summary(painters$School)
````

````r
# Summary of a single numeric variable
summary(painters$Composition)
````

## 9. `attach()` and `detach()`

````r
# Attach the data frame
attach(painters)

# Now columns can be referred to directly
summary(School)
summary(Composition)

# Always detach when finished
detach(painters)
````

After `detach(painters)`, the variables are no longer available as standalone names. The normal form is again:

```r
painters$School
painters$Composition
```

````r
# Safe/default form after detach
summary(painters$School)
summary(painters$Composition)
````

## 10. Subsetting a Data Frame

````r
# Select painters belonging to School F
subset(painters, School == "F")
````

````r
# Equivalent logical-indexing approach
painters[painters[["School"]] == "F", ]
````

````r
# Select painters with Composition <= 6
subset(painters, Composition <= 6)
````

### Remove unwanted columns with `select=`

````r
# Remove column 3 (Colour) and column 5 (School)
subset(
  painters,
  School == "F",
  select = c(-3, -5)
)
````

````r
# Instead of removing columns, explicitly select the required ones
subset(
  painters,
  School == "F",
  select = c(Composition, Drawing, Expression)
)
````

# 12. Combining Data Frames

The lecture gives three main techniques:

- `cbind()` → combine columns side-by-side
- `merge()` → join using a common column
- `rbind()` → stack rows vertically

## 13. `cbind()` – Combine Side-by-Side

````r
df1 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df2 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)

df1
df2
````

````r
# Horizontally combine the two data frames
cbind(df1, df2)
````

### Important

`cbind()` combines columns **by position/row**.

It does not use `state` to match rows.

## 14. `merge()` – Join Using a Common Column

````r
# Merge using the common variable 'state'
merged_df <- merge(df1, df2, by = "state")

merged_df
````

Here `state` is common to both data frames, so `merge()` matches records using `state`.

The lecture's example uses:

```r
merge(df1, df2, by = "state")
```

````r
# Explicitly inspect the result
str(merged_df)
````

### `merge()` with `by.x` and `by.y`

````r
# Demonstration where the common column has different names
left_df <- data.frame(
  state_code = c("UP", "MP", "AP", "JK"),
  population = c(1000, 2000, 3000, 4000)
)

right_df <- data.frame(
  state = c("AP", "JK", "MP", "UP"),
  capital = c("Amaravati", "Srinagar", "Bhopal", "Lucknow")
)

merge(
  left_df,
  right_df,
  by.x = "state_code",
  by.y = "state"
)
````

## 15. `rbind()` – Combine Vertically

````r
df11 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df22 <- data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)

df11
df22
````

````r
# Stack the second data frame below the first
rbind(df11, df22)
````

### Important condition for `rbind()`

The data frames should have compatible columns. Here both have:

- `state`
- `popnsize`

so the rows can be stacked.
