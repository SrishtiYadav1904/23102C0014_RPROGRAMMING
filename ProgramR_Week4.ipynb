## Lecture 16 – Logical Operators

Operators covered in the notes:
- `>` greater than
- `>=` greater than or equal to
- `<` less than
- `<=` less than or equal to
- `==` exactly equal to
- `!=` not equal to
- `!` logical NOT
- `&`, `&&` logical AND
- `|`, `||` logical OR
- `xor()` exclusive OR
- `isTRUE()`
- `isFALSE()`


### OR: `|` and `||`

```python
%%R
# Scalar example using ||
x = 8
(x < 10) || (x < 2)

x = 18
(x < 10) || (x < 2)
```

```python
%%R
# Vector example
x = c(8,18)

# || operates only on the first element
(x[1] < 10) || (x[1] < 2)

# | operates on all elements
(x < 10) | (x < 2)
```

### AND: `&` and `&&`

```python
%%R
# Scalar example using &&
x = 5
(x < 10) && (x > 2)

x = 15
(x < 10) && (x > 2)
```

```python
%%R
# Vector example
x = c(8,18)

# && operates only on the first element
(x[1] < 10) && (x[1] > 2)

# & operates on all elements
(x < 10) & (x > 2)
```

### Selecting values using a logical condition

```python
%%R
# Generate values from 1 to 6
x = 1:6

# Check which values are greater than 2 and less than 5
(x > 2) & (x < 5)

# Select values greater than 2 and less than 5
x[(x > 2) & (x < 5)]
```

### Logical OR with vectors

```python
%%R
# Generate values from 1 to 6
x = 1:6

# Check which values are greater than 2 or less than 5
(x > 2) | (x < 5)

# Select values satisfying the condition
x[(x > 2) | (x < 5)]
```

### Longer-form logical operators examine the first element

```python
%%R
# Generate values from 1 to 6
x = 1:6

# && evaluates using only the first element
(x[1] > 2) && (x[1] < 5)

# Equivalent first-element comparison
(x[1] > 2) & (x[1] < 5)
```
